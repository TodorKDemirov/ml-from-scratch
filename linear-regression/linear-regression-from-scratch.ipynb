In this notebook we will implement linear regression using only the most basic Python libraries and explaining the Math behind it.
Linear regression is a supervised machine learning method that models the linear relationship between a continuous dependant variable  $\mathcal{y}$ and one or more independent variables  $\mathcal{x}$.
We will implement it using three main functions:
##### 1. Cost Function
##### 2. Gradient Function

# Math explanation
The goal of linear regression is to find the best fitting hyperplane(which is a straight line in the 2D example) through some data points ($\mathcal{x}$,  $\mathcal{y}$), where  $\mathcal{x}$ represents one or more input features and  $\mathcal{y}$ is the corresponding target value. After training, the model is used to predict output values $\hat{y}$ for new unseen inputs $\mathcal{x}$. In the next image is shown how this looks on a 2D plane(so $\mathcal{x}$ is a single input feature).

<img src="lr-images/lr-example1.png" width="500">

This hyperplane can be represented as the function

$\mathcal{f(x) = w^T x + b}$,

where $\mathcal{w}$ is the vector of weights, with each weight representing the contribution of its corresponding feature to the prediction, and $\mathcal{b}$ is the intercept. These parameters will be used to predict $\hat{y}$.

$\mathcal{\hat{y}_i = w^T x_i + b}$

To find the optimal values for $\mathcal{w}$ and $\mathcal{b}$ we need to minimize the error between every predicted $\hat{y}_i$ and the actual value of $\mathcal{y}_i$. To do that we need the Cost function or MSE(mean squared error) which are represented as

$\mathcal{C(w,b) = \frac{1}{2m}\sum_{i=1}^{m} (f(x_i)-y_i)^2}$,

where $\mathcal{f(x_i)}$ is $\hat{y_i}$ and $\mathcal{f(x_i)-y_i}$ is the error of the model for the i-th observation. We square the error so that it is always positive and to penalize larger errors more heavily. Then we sum the se errors from 1 to $\mathcal{m}$ where $\mathcal{m}$ is the number of observations in the model so we can include the error from the whole model. Finally, we divide by $\mathcal{2m}$ to get the average. Dividing by $\mathcal{2m}$ rather than $\mathcal{m}$ makes the calculations cleaner. In the next image we can see the error shown more clearly.

<img src="lr-images/lr-example2.png" width="500">

To minimize the cost function we use Gradient Descent. For this algorithm we start with some initial $\mathcal{w}$ and $\mathcal{b}$(e.g. $\mathcal{w^T=(0,0,...,0), b=0}$) then we repeatedly update $\mathcal{w}$ and $\mathcal{b}$ in small steps to reduce the cost function. We continue until we reach the minimum of the function. The Gradient points in the direction of steepest ascent so each step we want to go in the opposite direction. We will now formally explain the Gradient Descent algorithm. At each iteration $\mathcal{w}$ and $\mathcal{b}$ are updated subtracting a small step $\alpha$ multiplied by the partial derivative with respect to $\mathcal{w}$ and $\mathcal{b}$ respectively.

$\mathcal{w}$ = $\mathcal{w}$ - $\alpha$ $\frac{\partial C(w,b)}{\partial w}$

$\mathcal{b}$ = $\mathcal{b}$ - $\alpha$ $\frac{\partial C(w,b)}{\partial b}$

$\alpha$ is called the learning rate. Making it too small will make the algorithm too slow because we will take only tiny steps. Making it too large might cause the algorithm to never converge to a minimum. So we need to choose it carefully.

This is how we will find the partial derivatives:

$\frac{\partial C(w,b)}{\partial w_j}$ = $\frac{\partial}{\partial w_j}$ $\frac{1}{2m}\sum_{i=1}^{m} (\sum_{k=1}^{n} w_k x_ik + b-y_i)^2 = \frac{1}{2m}\sum_{i=1}^{m} 2(\sum_{k=1}^{n} w_k x_{ik} + b-y_i)*\frac{\partial}{\partial w_j}(\sum_{k=1}^{n} w_k x_{ik} + b-y_i) = $

$= \frac{1}{2m}\sum_{i=1}^{m} 2(w^T x_i + b-y_i)*x_{ij} = \frac{1}{m}\sum_{i=1}^{m} x_{ij}(w^T x_i + b-y_i)$

$\frac{\partial C(w,b)}{\partial b}$ = $\frac{\partial}{\partial b}$ $\frac{1}{2m}\sum_{i=1}^{m} (w^T x_i + b-y_i)^2 = \frac{1}{2m}\sum_{i=1}^{m} 2(w^T x_i + b-y_i)*\frac{\partial }{\partial b}(w^T x_i + b-y_i) = $

$= \frac{1}{m}\sum_{i=1}^{m} (w^T x_i + b-y_i)$






































































```python
import numpy as np
```

```python
def cost_function(x, y, w, b):
    m = x.shape[0]  # number of rows (training examples)

    cost_sum = 0

    for i in range(m):
        f = x[i] @ w + b # the function for the predicted value
        error = (f - y[i]) ** 2 # calculating the prediction error
        cost_sum += error # taking the sum of the errors
    total_cost = cost_sum/(2*m) # taking the average of the errors
    return total_cost
```

```python
def gradient_function(x, y, w, b):
    m = x.shape[0] # number of rows (training examples)
    n = x.shape[1] # number of columns (features)
    dc_dw = np.zeros(n) # create a 0 vector for the derivatives of w1,...,wn
    dc_db = 0 # the derivative of b

    for i in range(m): # for loop for the sum from i to m
        f = x[i] @ w + b # the function for the predicted value
        for j in range(n):
            dc_dw[j] += x[i][j] * (f - y[i]) # calculating the derivatives of w
        dc_db += f - y[i] # calculating the derivative for b

    dc_dw = dc_dw/m # finishing the derivative calculations of w
    dc_db = dc_db/m # finishing the derivative calculations of b

    return dc_dw, dc_db
```

```python
def gradient_descent(x, y, alpha, iterations):
    n = x.shape[1] # number of columns (features)
    w = np.zeros(n) # creating a vector w
    b = 0
    for i in range(iterations): # looping over the algorithm for the number of iterations
        dc_dw, dc_db = gradient_function(x, y, w, b) # getting the derivatives
        w -= alpha * dc_dw # making a step for w
        b -= alpha * dc_db # making a step for b
        print(f"Iteration {i}: w = {w}, b = {b}, Cost = {cost_function(x, y, w, b)}")

    return w, b
```

```python
# TEST
# We are creating a little data sample where X contains 2 features.
# The expected result is for the model to find the correlation y = 2X + 1

X = np.array([[1.0], [2.0], [3.0], [4.0]])
y = np.array([3.0, 5.0, 7.0, 9.0])

w, b = gradient_descent(X, y, alpha=0.01, iterations=2000)

print("Learned weights:", w)
print("Learned bias:", b)
```
